```python
import os
import pandas as pd

base_path = "/root/.cache/kagglehub/datasets/ambarish/breakhis/versions/4/BreaKHis_v1/BreaKHis_v1/histology_slides/breast"

data = []

for root, dirs, files in os.walk(base_path):

    for file in files:

        if file.endswith(".png"):

            filepath = os.path.join(root, file)

            if "/benign/" in filepath:
                label = "benign"

            elif "/malignant/" in filepath:
                label = "malignant"

            else:
                continue

            data.append([filepath, label])

df = pd.DataFrame(data, columns=["filepath", "label"])

print(df.shape)
print(df["label"].value_counts())
```

Sample Histopathological Images

```python
from PIL import Image
import matplotlib.pyplot as plt

sample = df.sample(6)

plt.figure(figsize=(12,8))

for i, (_, row) in enumerate(sample.iterrows()):

    img = Image.open(row["filepath"])

    plt.subplot(2,3,i+1)
    plt.imshow(img)
    plt.title(row["label"])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    stratify=df['label'],
    random_state=42
)

print("Train:", train_df.shape)
print("Test:", test_df.shape)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_generator = train_datagen.flow_from_dataframe(
    train_df,
    x_col='filepath',
    y_col='label',
    target_size=(224,224),
    batch_size=16,
    class_mode='binary',
    subset='training'
)

val_generator = train_datagen.flow_from_dataframe(
    train_df,
    x_col='filepath',
    y_col='label',
    target_size=(224,224),
    batch_size=16,
    class_mode='binary',
    subset='validation'
)
```

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from tensorflow.keras.optimizers import Adam

base_model = ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)

base_model.trainable = False

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=Adam(0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
df_small = df.sample(n=500, random_state=42)

print(df_small.shape)
```

```python
train_df, test_df = train_test_split(
    df_small,
    test_size=0.2,
    stratify=df_small['label'],
    random_state=42
)
```

```python
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=2
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
test_datagen = ImageDataGenerator(rescale=1./255)

test_generator = test_datagen.flow_from_dataframe(
    test_df,
    x_col='filepath',
    y_col='label',
    target_size=(224,224),
    batch_size=16,
    class_mode='binary',
    shuffle=False
)
```

```python
test_loss, test_accuracy = model.evaluate(test_generator)

print("Test Accuracy:", test_accuracy)
print("Test Loss:", test_loss)
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns

predictions = model.predict(test_generator)

y_pred = (predictions > 0.5).astype(int)

cm = confusion_matrix(
    test_generator.classes,
    y_pred
)

plt.figure(figsize=(5,4))
sns.heatmap(cm,
            annot=True,
            fmt='d',
            cmap='Blues')

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        test_generator.classes,
        y_pred,
        target_names=['benign','malignant']
    )
)
```
